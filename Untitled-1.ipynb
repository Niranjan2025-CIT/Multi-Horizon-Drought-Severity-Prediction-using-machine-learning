```python
# ============================================================
# STEP 28 — CALCULATE NEIGHBOUR SCORE
# ============================================================

def calculate_neighbor_score(
    fips,
    drought_scores
):
    """
    Calculate the average drought score
    of the neighbouring counties.

    Parameters
    ----------
    fips : str
        Target county FIPS.

    drought_scores : dict
        Dictionary:
            FIPS -> temporary 0–5 score

    Returns
    -------
    float
        Average neighbour drought score.
    """

    fips = str(fips).zfill(5)

    neighbour_list = neighbors.get(
        fips,
        []
    )

    if len(neighbour_list) == 0:
        return np.nan

    scores = []

    for neighbour_fips in neighbour_list:

        neighbour_fips = str(
            neighbour_fips
        ).zfill(5)

        if neighbour_fips in drought_scores:

            score = drought_scores[
                neighbour_fips
            ]

            if not pd.isna(score):
                scores.append(score)

    if len(scores) == 0:
        return np.nan

    return float(
        np.mean(scores)
    )
```

```python
# ============================================================
# STEP 32 — CLEAN USDM DROUGHT COLUMNS
# ============================================================

DROUGHT_COLUMNS = [
    "D0",
    "D1",
    "D2",
    "D3",
    "D4"
]

# Convert all drought-area columns to numeric
for col in DROUGHT_COLUMNS:

    usdm_df[col] = pd.to_numeric(
        usdm_df[col],
        errors="coerce"
    )


print("USDM drought column types:")
print(
    usdm_df[DROUGHT_COLUMNS].dtypes
)
```

```python
# ============================================================
# STEP 32 — USDM ROW → DROUGHT CLASS
# ============================================================

DROUGHT_COLUMNS = [
    "D0",
    "D1",
    "D2",
    "D3",
    "D4"
]


def usdm_row_to_class(row):
    """
    Convert one USDM county row into
    a single drought class.

    Temporary mapping:

    No drought -> 0
    D0         -> 1
    D1         -> 2
    D2         -> 3
    D3         -> 4
    D4         -> 5
    """

    # Check from most severe to least severe
    for class_number, drought_column in reversed(
        list(enumerate(DROUGHT_COLUMNS, start=1))
    ):

        # Explicitly convert API value to numeric
        value = pd.to_numeric(
            row[drought_column],
            errors="coerce"
        )

        if pd.notna(value) and value > 0:

            return class_number

    # No D0-D4 area
    return 0
```

```python
# ============================================================
# STEP 34 — FETCH COUNTY + NEIGHBOUR DROUGHT SCORES
# ============================================================

def get_usdm_score_for_fips(
    fips,
    start_date="01/01/2026",
    end_date="08/08/2026"
):
    """
    Fetch the latest USDM drought condition
    for one county.

    Returns:
        drought_class
        temporary 0–5 drought score
        map date
    """

    fips = str(fips).zfill(5)

    params = {
        "aoi": fips,
        "startdate": start_date,
        "enddate": end_date,
        "statisticsType": "1"
    }

    response = requests.get(
        USDM_URL,
        params=params,
        timeout=30
    )

    response.raise_for_status()

    df = pd.read_csv(
        io.StringIO(response.text)
    )

    if df.empty:
        return {
            "fips": fips,
            "drought_class": np.nan,
            "score": np.nan,
            "map_date": None
        }

    # --------------------------------------------------------
    # Clean D0-D4
    # --------------------------------------------------------

    for col in DROUGHT_COLUMNS:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    # --------------------------------------------------------
    # Latest available USDM map
    # --------------------------------------------------------

    df = df.sort_values(
        "MapDate",
        ascending=False
    )

    latest = df.iloc[0]

    # --------------------------------------------------------
    # Convert to class
    # --------------------------------------------------------

    drought_class = (
        usdm_row_to_class(latest)
    )

    # --------------------------------------------------------
    # Convert class to temporary 0–5 score
    # --------------------------------------------------------

    drought_score = (
        drought_class_to_score(
            drought_class
        )
    )

    return {
        "fips": fips,
        "drought_class": drought_class,
        "score": drought_score,
        "map_date": latest["MapDate"]
    }
```

```python
# ============================================================
# STEP 35A — DEFINE USDM API URL
# ============================================================

import requests
import pandas as pd
import numpy as np
import io

USDM_URL = (
    "https://usdmdataservices.unl.edu/"
    "api/CountyStatistics/"
    "GetDroughtSeverityStatisticsByArea"
)

DROUGHT_COLUMNS = [
    "D0",
    "D1",
    "D2",
    "D3",
    "D4"
]

print("USDM API configured.")
```

```python
# ============================================================
# STEP 35B — COUNTY DROUGHT SCORE FUNCTION
# ============================================================

def drought_class_to_score(drought_class):

    if pd.isna(drought_class):
        return np.nan

    drought_class = int(drought_class)

    return drought_class * (5 / 4)


def usdm_row_to_class(row):

    # D4 → D3 → D2 → D1 → D0
    for class_number, drought_column in reversed(
        list(enumerate(DROUGHT_COLUMNS, start=1))
    ):

        value = pd.to_numeric(
            row[drought_column],
            errors="coerce"
        )

        if pd.notna(value) and value > 0:
            return class_number

    return 0


def get_usdm_score_for_fips(
    fips,
    start_date="01/01/2026",
    end_date="08/08/2026"
):

    fips = str(fips).zfill(5)

    params = {
        "aoi": fips,
        "startdate": start_date,
        "enddate": end_date,
        "statisticsType": "1"
    }

    response = requests.get(
        USDM_URL,
        params=params,
        timeout=30
    )

    print(
        f"USDM request: {fips} | "
        f"HTTP {response.status_code}"
    )

    response.raise_for_status()

    df = pd.read_csv(
        io.StringIO(response.text)
    )

    if df.empty:

        return {
            "fips": fips,
            "drought_class": np.nan,
            "score": np.nan,
            "map_date": None
        }

    # --------------------------------------------------------
    # Convert D0-D4 columns to numeric
    # --------------------------------------------------------

    for col in DROUGHT_COLUMNS:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    # --------------------------------------------------------
    # Latest available USDM week
    # --------------------------------------------------------

    df = df.sort_values(
        "MapDate",
        ascending=False
    )

    latest = df.iloc[0]

    # --------------------------------------------------------
    # Determine drought class
    # --------------------------------------------------------

    drought_class = usdm_row_to_class(
        latest
    )

    # --------------------------------------------------------
    # Temporary 0–5 conversion
    # --------------------------------------------------------

    drought_score = drought_class_to_score(
        drought_class
    )

    return {
        "fips": fips,
        "drought_class": drought_class,
        "score": drought_score,
        "map_date": latest["MapDate"]
    }
```

```python
# ============================================================
# STEP 36 — FETCH NEIGHBOUR DROUGHT SCORES
# ============================================================

def get_neighbor_scores(fips):

    fips = str(fips).zfill(5)

    # --------------------------------------------------------
    # Get neighbours from saved dictionary
    # --------------------------------------------------------

    neighbour_list = neighbors.get(
        fips,
        []
    )

    print(
        f"County {fips} has "
        f"{len(neighbour_list)} neighbours."
    )

    if len(neighbour_list) == 0:

        print(
            "WARNING: No neighbours found."
        )

        return []

    results = []

    # --------------------------------------------------------
    # Fetch USDM score for every neighbour
    # --------------------------------------------------------

    for i, neighbour_fips in enumerate(
        neighbour_list,
        start=1
    ):

        neighbour_fips = str(
            neighbour_fips
        ).zfill(5)

        print(
            f"[{i}/{len(neighbour_list)}] "
            f"Fetching {neighbour_fips}..."
        )

        try:

            result = get_usdm_score_for_fips(
                neighbour_fips
            )

            results.append(result)

        except Exception as e:

            print(
                f"ERROR for {neighbour_fips}: {e}"
            )

            results.append({
                "fips": neighbour_fips,
                "drought_class": np.nan,
                "score": np.nan,
                "map_date": None
            })

    return results
```

```python
# ============================================================
# STEP 37 — CALCULATE NEIGHBOUR SCORE
# ============================================================

def get_neighbor_score(fips):

    results = get_neighbor_scores(
        fips
    )

    valid_scores = []

    for result in results:

        score = result["score"]

        if pd.notna(score):

            valid_scores.append(
                float(score)
            )

    # --------------------------------------------------------
    # No valid neighbour scores
    # --------------------------------------------------------

    if len(valid_scores) == 0:

        return np.nan

    # --------------------------------------------------------
    # Average neighbour score
    # --------------------------------------------------------

    neighbor_score = np.mean(
        valid_scores
    )

    return float(
        neighbor_score
    )
```

```python
# ============================================================
# STEP 41 — CALCULATE WEATHER FEATURES
# ============================================================

def calculate_weather_features(weather_df):

    df = weather_df.copy()

    # --------------------------------------------------------
    # Check required raw columns
    # --------------------------------------------------------

    required_columns = [
        "date",
        "T2M",
        "QV2M",
        "WS10M",
        "PRECTOT"
    ]

    missing = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing weather columns: {missing}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    # --------------------------------------------------------
    # Date
    # --------------------------------------------------------

    df["date"] = pd.to_datetime(
        df["date"],
        errors="coerce"
    )

    df = df.sort_values(
        "date"
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Convert weather columns to numeric
    # --------------------------------------------------------

    weather_columns = [
        "T2M",
        "QV2M",
        "WS10M",
        "PRECTOT"
    ]

    for col in weather_columns:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    # --------------------------------------------------------
    # NASA invalid values
    # --------------------------------------------------------

    df[weather_columns] = (
        df[weather_columns]
        .replace(-999, np.nan)
    )

    # --------------------------------------------------------
    # 7-day precipitation mean
    # --------------------------------------------------------

    df["PRECTOT_mean_7d"] = (
        df["PRECTOT"]
        .rolling(
            window=7,
            min_periods=7
        )
        .mean()
    )

    # --------------------------------------------------------
    # 30-day precipitation mean
    # --------------------------------------------------------

    df["PRECTOT_mean_30d"] = (
        df["PRECTOT"]
        .rolling(
            window=30,
            min_periods=30
        )
        .mean()
    )

    # --------------------------------------------------------
    # 30-day temperature mean
    # --------------------------------------------------------

    df["T2M_mean_30d"] = (
        df["T2M"]
        .rolling(
            window=30,
            min_periods=30
        )
        .mean()
    )

    # --------------------------------------------------------
    # 30-day humidity mean
    # --------------------------------------------------------

    df["QV2M_mean_30d"] = (
        df["QV2M"]
        .rolling(
            window=30,
            min_periods=30
        )
        .mean()
    )

    # --------------------------------------------------------
    # 30-day wind-speed mean
    # --------------------------------------------------------

    df["WS10M_mean_30d"] = (
        df["WS10M"]
        .rolling(
            window=30,
            min_periods=30
        )
        .mean()
    )

    return df
```

```python
# ============================================================
# STEP 46 — NASA WEATHER FETCH FUNCTION
# ============================================================

import requests
import pandas as pd
import numpy as np


# ============================================================
# STEP 46 — NASA WEATHER FETCH FUNCTION
# ============================================================

def get_weather_data(
    fips,
    start_date="2026-07-01",
    end_date="2026-08-08"
):

    fips = str(fips).zfill(5)

    # --------------------------------------------------------
    # Get county coordinates
    # --------------------------------------------------------

    county_match = county_locations[
        county_locations["fips"]
        .astype(str)
        .str.zfill(5)
        == fips
    ]

    if county_match.empty:

        raise ValueError(
            f"FIPS {fips} not found in county_locations."
        )

    county = county_match.iloc[0]

    latitude = float(
        county["latitude"]
    )

    longitude = float(
        county["longitude"]
    )

    # --------------------------------------------------------
    # NASA POWER API
    # --------------------------------------------------------

    url = (
        "https://power.larc.nasa.gov/"
        "api/temporal/daily/point"
    )

    params = {

        "parameters": (
            "T2M,"
            "T2M_MAX,"
            "T2M_MIN,"
            "QV2M,"
            "WS10M,"
            "PRECTOTCORR"
        ),

        "community": "AG",

        "longitude": longitude,

        "latitude": latitude,

        "start": start_date.replace("-", ""),

        "end": end_date.replace("-", ""),

        "format": "JSON",

        "time-standard": "UTC"
    }

    # --------------------------------------------------------
    # Request NASA
    # --------------------------------------------------------

    response = requests.get(
        url,
        params=params,
        timeout=60
    )

    response.raise_for_status()

    data = response.json()

    # --------------------------------------------------------
    # Extract weather
    # --------------------------------------------------------

    parameter_data = (
        data["properties"]["parameter"]
    )

    weather = pd.DataFrame(
        parameter_data
    )

    weather.index.name = "date"

    weather = (
        weather
        .reset_index()
    )

    # --------------------------------------------------------
    # Convert date
    # --------------------------------------------------------

    weather["date"] = pd.to_datetime(
        weather["date"],
        format="%Y%m%d",
        errors="coerce"
    )

    # --------------------------------------------------------
    # IMPORTANT:
    # Rename NASA PRECTOTCORR → PRECTOT
    #
    # This makes the live data match the
    # feature name used during model training.
    # --------------------------------------------------------

    weather = weather.rename(
        columns={
            "PRECTOTCORR": "PRECTOT"
        }
    )

    # --------------------------------------------------------
    # Convert numeric columns
    # --------------------------------------------------------

    numeric_columns = [
        "T2M",
        "T2M_MAX",
        "T2M_MIN",
        "QV2M",
        "WS10M",
        "PRECTOT"
    ]

    for col in numeric_columns:

        weather[col] = pd.to_numeric(
            weather[col],
            errors="coerce"
        )

        # NASA invalid values
        weather.loc[
            weather[col] <= -900,
            col
        ] = np.nan

    # --------------------------------------------------------
    # Sort
    # --------------------------------------------------------

    weather = (
        weather
        .sort_values("date")
        .reset_index(drop=True)
    )

    return weather
```

```python
# ============================================================
# STEP 49 — ROBUST NEIGHBOUR WEATHER FEATURES
# ============================================================

NEIGHBOR_WEATHER_FEATURES = [
    "PRECTOT_mean_7d",
    "PRECTOT_mean_30d",
    "T2M_mean_30d",
    "QV2M_mean_30d",
    "WS10M_mean_30d"
]


def get_neighbor_weather_features(fips):

    fips = str(fips).zfill(5)

    neighbour_list = neighbors.get(
        fips,
        []
    )

    if len(neighbour_list) == 0:

        raise ValueError(
            f"No neighbours found for {fips}"
        )

    all_features = []

    print("=" * 60)
    print(
        f"COUNTY: {fips}"
    )
    print(
        f"NEIGHBOURS: {len(neighbour_list)}"
    )
    print("=" * 60)

    for i, neighbour_fips in enumerate(
        neighbour_list,
        start=1
    ):

        neighbour_fips = str(
            neighbour_fips
        ).zfill(5)

        print(
            f"\n[{i}/{len(neighbour_list)}] "
            f"{neighbour_fips}"
        )

        try:

            # ------------------------------------------------
            # Fetch weather
            # ------------------------------------------------

            raw_weather = get_weather_data(
                neighbour_fips
            )

            # ------------------------------------------------
            # IMPORTANT DEBUG CHECK
            # ------------------------------------------------

            print(
                "   Returned type:",
                type(raw_weather).__name__
            )

            if not isinstance(
                raw_weather,
                pd.DataFrame
            ):

                raise TypeError(
                    "get_weather_data() did not "
                    "return a DataFrame."
                )

            # ------------------------------------------------
            # Check PRECTOT
            # ------------------------------------------------

            if "PRECTOT" not in raw_weather.columns:

                # Handle NASA name if necessary
                if "PRECTOTCORR" in raw_weather.columns:

                    raw_weather = raw_weather.rename(
                        columns={
                            "PRECTOTCORR": "PRECTOT"
                        }
                    )

                else:

                    raise ValueError(
                        "Weather data has neither "
                        "PRECTOT nor PRECTOTCORR."
                    )

            # ------------------------------------------------
            # Calculate rolling features
            # ------------------------------------------------

            features = (
                calculate_weather_features(
                    raw_weather
                )
            )

            # ------------------------------------------------
            # Find latest valid row
            # ------------------------------------------------

            valid = (
                features
                .dropna(
                    subset=NEIGHBOR_WEATHER_FEATURES
                )
            )

            if valid.empty:

                print(
                    "   No valid 30-day "
                    "weather features."
                )

                continue

            latest = (
                valid
                .sort_values("date")
                .iloc[-1]
            )

            # ------------------------------------------------
            # Store features
            # ------------------------------------------------

            row = {}

            for feature in (
                NEIGHBOR_WEATHER_FEATURES
            ):

                row[feature] = float(
                    latest[feature]
                )

            all_features.append(
                row
            )

            print(
                "   Latest valid date:",
                latest["date"]
            )

            print(
                "   Features collected."
            )

        except Exception as e:

            print(
                "   FAILED:",
                repr(e)
            )

    # --------------------------------------------------------
    # Make sure we got something
    # --------------------------------------------------------

    if not all_features:

        raise ValueError(
            "Could not obtain valid weather "
            "features for any neighbour."
        )

    # --------------------------------------------------------
    # Convert to DataFrame
    # --------------------------------------------------------

    neighbour_weather_df = pd.DataFrame(
        all_features
    )

    print()
    print("=" * 60)
    print("VALID NEIGHBOURS:", len(
        neighbour_weather_df
    ))
    print("=" * 60)

    print(
        neighbour_weather_df
    )

    # --------------------------------------------------------
    # Average
    # --------------------------------------------------------

    neighbour_mean = (
        neighbour_weather_df.mean(
            numeric_only=True
        )
    )

    return neighbour_mean
```

```python
# ============================================================
# STEP 51 — BUILD COMPLETE 24-FEATURE VECTOR
# ============================================================

MODEL_FEATURES = [
    "score",
    "neighbor_score",
    "PRECTOT_mean_7d",
    "PRECTOT_mean_30d",
    "neighbor_PRECTOT_mean_7d",
    "neighbor_PRECTOT_mean_30d",
    "T2M_mean_30d",
    "neighbor_T2M_mean_30d",
    "QV2M_mean_30d",
    "neighbor_QV2M_mean_30d",
    "WS10M_mean_30d",
    "neighbor_WS10M_mean_30d",
    "month_sin",
    "month_cos",
    "day_sin",
    "day_cos",
    "CULT_LAND",
    "SQ1",
    "SQ2",
    "SQ3",
    "SQ4",
    "SQ5",
    "SQ6",
    "SQ7"
]


def build_prediction_features(fips):

    fips = str(fips).zfill(5)

    print("=" * 70)
    print(
        f"BUILDING FEATURES FOR {fips}"
    )
    print("=" * 70)

    # ========================================================
    # 1. CURRENT USDM SCORE
    # ========================================================

    current_result = (
        get_usdm_score_for_fips(
            fips
        )
    )

    if current_result is None:

        raise ValueError(
            f"No USDM result for {fips}"
        )

    # get_usdm_score_for_fips()
    # returns a dictionary

    current_score = current_result[
        "score"
    ]

    current_score = float(
        current_score
    )

    print(
        "\nCurrent drought score:",
        current_score
    )

    # ========================================================
    # 2. NEIGHBOUR SCORE
    # ========================================================

    neighbour_list = neighbors.get(
        fips,
        []
    )

    if not neighbour_list:

        raise ValueError(
            f"No neighbours found for {fips}"
        )

    neighbour_scores = []

    print(
        "\nNeighbour scores:"
    )

    for neighbour_fips in neighbour_list:

        neighbour_fips = str(
            neighbour_fips
        ).zfill(5)

        try:

            result = (
                get_usdm_score_for_fips(
                    neighbour_fips
                )
            )

            if result is None:

                print(
                    f"{neighbour_fips}: "
                    "No result"
                )

                continue

            # ------------------------------------------------
            # Extract numeric score from dictionary
            # ------------------------------------------------

            score = result[
                "score"
            ]

            score = float(
                score
            )

            neighbour_scores.append(
                score
            )

            print(
                f"{neighbour_fips}: "
                f"{score}"
            )

        except Exception as e:

            print(
                f"{neighbour_fips}: "
                f"ERROR - {e}"
            )

    if not neighbour_scores:

        raise ValueError(
            "No valid neighbour scores."
        )

    neighbour_score = np.mean(
        neighbour_scores
    )

    print(
        "\nNeighbour average:",
        neighbour_score
    )

    # ========================================================
    # 3. CURRENT WEATHER
    # ========================================================

    current_weather = (
        get_weather_data(
            fips
        )
    )

    current_weather_features = (
        calculate_weather_features(
            current_weather
        )
    )

    required_weather = [
        "PRECTOT_mean_7d",
        "PRECTOT_mean_30d",
        "T2M_mean_30d",
        "QV2M_mean_30d",
        "WS10M_mean_30d"
    ]

    valid_current = (
        current_weather_features
        .dropna(
            subset=required_weather
        )
    )

    if valid_current.empty:

        raise ValueError(
            "No valid current weather "
            "features."
        )

    latest_current = (
        valid_current
        .sort_values("date")
        .iloc[-1]
    )

    latest_date = pd.to_datetime(
        latest_current["date"]
    )

    print(
        "\nLatest weather date:",
        latest_date
    )

    # ========================================================
    # 4. NEIGHBOUR WEATHER
    # ========================================================

    neighbour_weather = (
        get_neighbor_weather_features(
            fips
        )
    )

    # ========================================================
    # 5. CYCLICAL DATE FEATURES
    # ========================================================

    month = latest_date.month

    day_of_year = (
        latest_date.dayofyear
    )

    month_sin = np.sin(
        2 * np.pi * month / 12
    )

    month_cos = np.cos(
        2 * np.pi * month / 12
    )

    day_sin = np.sin(
        2 * np.pi * day_of_year / 365
    )

    day_cos = np.cos(
        2 * np.pi * day_of_year / 365
    )

    # ========================================================
    # 6. SOIL DATA
    # ========================================================

    soil_row = soil_features[
        soil_features["fips"]
        .astype(str)
        .str.zfill(5)
        == fips
    ]

    if soil_row.empty:

        raise ValueError(
            f"No soil data found for {fips}"
        )

    soil_row = (
        soil_row.iloc[0]
    )

    # ========================================================
    # 7. BUILD FEATURE DICTIONARY
    # ========================================================

    feature_dict = {

        "score":
            current_score,

        "neighbor_score":
            float(neighbour_score),

        "PRECTOT_mean_7d":
            float(
                latest_current[
                    "PRECTOT_mean_7d"
                ]
            ),

        "PRECTOT_mean_30d":
            float(
                latest_current[
                    "PRECTOT_mean_30d"
                ]
            ),

        "neighbor_PRECTOT_mean_7d":
            float(
                neighbour_weather[
                    "PRECTOT_mean_7d"
                ]
            ),

        "neighbor_PRECTOT_mean_30d":
            float(
                neighbour_weather[
                    "PRECTOT_mean_30d"
                ]
            ),

        "T2M_mean_30d":
            float(
                latest_current[
                    "T2M_mean_30d"
                ]
            ),

        "neighbor_T2M_mean_30d":
            float(
                neighbour_weather[
                    "T2M_mean_30d"
                ]
            ),

        "QV2M_mean_30d":
            float(
                latest_current[
                    "QV2M_mean_30d"
                ]
            ),

        "neighbor_QV2M_mean_30d":
            float(
                neighbour_weather[
                    "QV2M_mean_30d"
                ]
            ),

        "WS10M_mean_30d":
            float(
                latest_current[
                    "WS10M_mean_30d"
                ]
            ),

        "neighbor_WS10M_mean_30d":
            float(
                neighbour_weather[
                    "WS10M_mean_30d"
                ]
            ),

        "month_sin":
            float(month_sin),

        "month_cos":
            float(month_cos),

        "day_sin":
            float(day_sin),

        "day_cos":
            float(day_cos)
    }

    # ========================================================
    # 8. SOIL FEATURES
    # ========================================================

    soil_columns = [
        "CULT_LAND",
        "SQ1",
        "SQ2",
        "SQ3",
        "SQ4",
        "SQ5",
        "SQ6",
        "SQ7"
    ]

    for col in soil_columns:

        feature_dict[col] = float(
            soil_row[col]
        )

    # ========================================================
    # 9. CREATE DATAFRAME
    # ========================================================

    X_live = pd.DataFrame(
        [feature_dict]
    )

    # ========================================================
    # 10. FORCE EXACT MODEL FEATURE ORDER
    # ========================================================

    X_live = X_live[
        MODEL_FEATURES
    ]

    # ========================================================
    # 11. FINAL CHECK
    # ========================================================

    print()
    print("=" * 70)
    print(
        "FEATURE VECTOR CREATED"
    )
    print("=" * 70)

    print(
        "Shape:",
        X_live.shape
    )

    print(
        "Features:",
        len(X_live.columns)
    )

    return X_live
```

```python
# ============================================================
# STEP 57 — COMPLETE COUNTY PREDICTION
# ============================================================

def predict_county(fips):

    fips = str(fips).zfill(5)

    print("=" * 70)
    print(f"PREDICTING DROUGHT FOR COUNTY: {fips}")
    print("=" * 70)

    # --------------------------------------------------------
    # Build the 24 features
    # --------------------------------------------------------

    X_live = build_prediction_features(
        fips
    )

    # --------------------------------------------------------
    # Predictions
    # --------------------------------------------------------

    predictions = {}

    for week in range(1, 5):

        model_path = os.path.join(
            "models",
            f"model_{week}w.pkl"
        )

        model = joblib.load(
            model_path
        )

        pred = model.predict(
            X_live
        )[0]

        predictions[
            f"pred_{week}w"
        ] = int(pred)

    # --------------------------------------------------------
    # Result
    # --------------------------------------------------------

    result = {
        "fips": fips,
        **predictions
    }

    return result
```
