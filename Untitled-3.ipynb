```python
import geopandas as gpd
import pandas as pd
import os

os.makedirs("data", exist_ok=True)

# Load Texas GeoJSON
counties = gpd.read_file("data/texas_geojson.json")

# Keep only FIPS
counties["fips"] = counties["id"]

# Convert to WGS84 (latitude/longitude)
counties = counties.to_crs(epsg=4326)

# Compute centroid
counties["centroid"] = counties.geometry.centroid

counties["latitude"] = counties.centroid.y
counties["longitude"] = counties.centroid.x

county_locations = counties[
    [
        "fips",
        "latitude",
        "longitude"
    ]
]

county_locations.to_csv(
    "data/texas_county_locations.csv",
    index=False
)

print(county_locations.head())
print()
print("Saved:", len(county_locations), "Texas counties")
```

```python
# ============================================================
# CALCULATE COUNTY CENTROIDS CORRECTLY
# ============================================================

# Project to a suitable projected CRS
counties_projected = counties.to_crs(epsg=3081)

# Calculate centroid in projected CRS
counties_projected["centroid"] = counties_projected.geometry.centroid

# Convert centroid back to WGS84
centroids = counties_projected["centroid"].to_crs(epsg=4326)

county_locations = pd.DataFrame({
    "fips": counties_projected["fips"].values,
    "latitude": centroids.y.values,
    "longitude": centroids.x.values
})

county_locations.to_csv(
    "data/texas_county_locations.csv",
    index=False
)

print(county_locations.head())
print()
print("Saved:", len(county_locations), "Texas counties")
```

```python
def predict_county(fips):

    fips = str(fips).zfill(5)

    print("=" * 70)
    print(f"PREDICTING DROUGHT FOR COUNTY: {fips}")
    print("=" * 70)

    X_live = build_prediction_features(fips)

    predictions = {}

    for week in range(1, 5):

        model_path = os.path.join(
            "models",
            f"model_{week}w.pkl"
        )

        model = joblib.load(model_path)

        pred = model.predict(X_live)[0]

        predictions[f"pred_{week}w"] = int(pred)

    return {
        "fips": fips,
        **predictions
    }
```
