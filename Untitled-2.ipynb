```python
# ============================================================
# NOTEBOOK 1 - SECTION 1
# LOAD DATA AND MERGE SOIL INFORMATION
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

data_dir = r"C:\Users\niran\Downloads\Drought prediction_ML\data"

# ------------------------------------------------------------
# Weather columns
# ------------------------------------------------------------

weather_cols = [
    "fips",
    "date",
    "PRECTOT",
    "QV2M",
    "T2M",
    "T2M_MAX",
    "T2M_MIN",
    "WS10M",
    "score"
]

# ------------------------------------------------------------
# Read datasets
# ------------------------------------------------------------

train = pd.read_csv(
    f"{data_dir}/train_timeseries.csv",
    usecols=weather_cols
)

val = pd.read_csv(
    f"{data_dir}/validation_timeseries.csv",
    usecols=weather_cols
)

test = pd.read_csv(
    f"{data_dir}/test_timeseries.csv",
    usecols=weather_cols
)

soil = pd.read_csv(
    f"{data_dir}/soil_data.csv"
)

print("Weather Shape :")
print(train.shape, val.shape, test.shape)

print("\nSoil Shape :")
print(soil.shape)

# ------------------------------------------------------------
# Preserve split information
# ------------------------------------------------------------

train["split"] = "train"
val["split"] = "validation"
test["split"] = "test"

# ------------------------------------------------------------
# Combine datasets
# ------------------------------------------------------------

full = pd.concat(
    [train, val, test],
    ignore_index=True
)

print("\nCombined Shape :", full.shape)

# ------------------------------------------------------------
# Format FIPS
# ------------------------------------------------------------

full["fips"] = full["fips"].astype(str).str.zfill(5)
soil["fips"] = soil["fips"].astype(str).str.zfill(5)

# ------------------------------------------------------------
# Keep Texas counties only
# ------------------------------------------------------------

full = full[
    full["fips"].str.startswith("48")
].copy()

soil = soil[
    soil["fips"].str.startswith("48")
].copy()

print("\nTexas Rows :", full.shape)
print("Texas Counties :", full["fips"].nunique())

# ------------------------------------------------------------
# Remove duplicate soil rows if any
# ------------------------------------------------------------

soil = soil.drop_duplicates("fips")

# ------------------------------------------------------------
# Merge soil
# ------------------------------------------------------------

df = full.merge(
    soil,
    on="fips",
    how="left"
)

# ------------------------------------------------------------
# Convert date
# ------------------------------------------------------------

df["date"] = pd.to_datetime(df["date"])

# ------------------------------------------------------------
# Sort
# ------------------------------------------------------------

df = (
    df
    .sort_values(["fips", "date"])
    .reset_index(drop=True)
)

print("\nFinal Shape :", df.shape)

print("\nMissing Soil Rows :")
print(df["SQ1"].isna().sum())

print("\nColumns :")
print(df.columns.tolist())

print("\nSample :")
print(df.head())
```

```python
# ============================================================
# NOTEBOOK 1 - SECTION 2
# FEATURE ENGINEERING
# ============================================================

import numpy as np

# ------------------------------------------------------------
# Calendar Features
# ------------------------------------------------------------

df["month"] = df["date"].dt.month
df["day_of_year"] = df["date"].dt.dayofyear

# Cyclic Encoding

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

df["day_sin"] = np.sin(2 * np.pi * df["day_of_year"] / 365.25)
df["day_cos"] = np.cos(2 * np.pi * df["day_of_year"] / 365.25)

# ------------------------------------------------------------
# Rolling Weather Features
# ------------------------------------------------------------

weather_vars = [
    "PRECTOT",
    "T2M",
    "T2M_MAX",
    "T2M_MIN",
    "QV2M",
    "WS10M"
]

windows = [7, 14, 30, 60]

for var in weather_vars:

    print("Processing", var)

    for w in windows:

        df[f"{var}_mean_{w}d"] = (
            df.groupby("fips")[var]
              .transform(
                  lambda x: x.rolling(
                      window=w,
                      min_periods=1
                  ).mean()
              )
        )

# ------------------------------------------------------------
# Rainfall Sums
# ------------------------------------------------------------

for w in [7, 14, 30]:

    df[f"PRECTOT_sum_{w}d"] = (
        df.groupby("fips")["PRECTOT"]
          .transform(
              lambda x: x.rolling(
                  window=w,
                  min_periods=1
              ).sum()
          )
    )

# ------------------------------------------------------------
# Weather Anomalies
# ------------------------------------------------------------

df["PRECTOT_anomaly"] = (
    df["PRECTOT_mean_7d"]
    -
    df["PRECTOT_mean_30d"]
)

df["T2M_anomaly"] = (
    df["T2M_mean_7d"]
    -
    df["T2M_mean_30d"]
)

# ------------------------------------------------------------
# Temperature Range
# ------------------------------------------------------------

df["TEMP_RANGE"] = (
    df["T2M_MAX"]
    -
    df["T2M_MIN"]
)

# ------------------------------------------------------------
# Check
# ------------------------------------------------------------

print(df.shape)

new_cols = [
    c for c in df.columns
    if "mean" in c
    or "sum" in c
    or "anomaly" in c
]

print("\nNumber of new weather features :", len(new_cols))

print(df[new_cols].head())
```

```python
# ============================================================
# NOTEBOOK 1 - SECTION 3
# WEEKLY DATASET + PREDICTION TARGETS
# ============================================================

# ------------------------------------------------------------
# Keep only rows where drought score exists
# ------------------------------------------------------------

weekly_df = df[df["score"].notna()].copy()

print("Weekly observations :", weekly_df.shape)

# ------------------------------------------------------------
# Sort
# ------------------------------------------------------------

weekly_df = (
    weekly_df
    .sort_values(["fips", "date"])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Create future targets
# ------------------------------------------------------------

for horizon in [1, 2, 3, 4]:

    weekly_df[f"target_{horizon}w"] = (
        weekly_df
        .groupby("fips")["score"]
        .shift(-horizon)
    )

# ------------------------------------------------------------
# Weeks since first observation
# ------------------------------------------------------------

weekly_df["week_number"] = (
    weekly_df
    .groupby("fips")
    .cumcount()
)

# ------------------------------------------------------------
# Check missing targets
# ------------------------------------------------------------

print("\nMissing Targets")

for h in [1,2,3,4]:

    print(
        f"target_{h}w :",
        weekly_df[f"target_{h}w"].isna().sum()
    )

print("\nDataset Shape")
print(weekly_df.shape)

# ------------------------------------------------------------
# Sample
# ------------------------------------------------------------

print(

weekly_df[
[
"fips",
"date",
"score",
"target_1w",
"target_2w",
"target_3w",
"target_4w",
"split"
]
].head(15)

)
```

```python
# ============================================================
# NOTEBOOK 1 - SECTION 4
# SPATIAL (NEIGHBOR) FEATURES
# ============================================================

import numpy as np

# ------------------------------------------------------------
# Load adjacency matrix
# ------------------------------------------------------------

A = np.load("texas_adjacency.npy")

county_order = (
    pd.read_csv("texas_county_order.csv")["fips"]
    .astype(str)
    .str.zfill(5)
    .tolist()
)

county_to_idx = {
    fips: i
    for i, fips in enumerate(county_order)
}

# ------------------------------------------------------------
# Map counties to matrix rows
# ------------------------------------------------------------

weekly_df["fips"] = weekly_df["fips"].astype(str).str.zfill(5)

weekly_df["county_idx"] = (
    weekly_df["fips"]
    .map(county_to_idx)
)

# Remove counties not present in adjacency matrix
weekly_df = weekly_df[
    weekly_df["county_idx"].notna()
].copy()

weekly_df["county_idx"] = (
    weekly_df["county_idx"]
    .astype(int)
)

# ------------------------------------------------------------
# Sort
# ------------------------------------------------------------

weekly_df = (
    weekly_df
    .sort_values(["date", "county_idx"])
    .reset_index(drop=True) a:[b,c,d]
)

# ------------------------------------------------------------
# Features whose neighbour average we want
# ------------------------------------------------------------

neighbor_features = [

    "score",

    "PRECTOT_mean_7d",
    "PRECTOT_mean_30d",

    "T2M_mean_30d",

    "QV2M_mean_30d",

    "WS10M_mean_30d"

]

# ------------------------------------------------------------
# Build neighbour features
# ------------------------------------------------------------

neighbor_tables = []

dates = np.sort(
    weekly_df["date"].unique()
)

for date in dates:

    week = (
        weekly_df[
            weekly_df["date"] == date
        ]
        .sort_values("county_idx")
        .copy()
    )

    # Ensure counties align with adjacency matrix
    week = (
        week
        .set_index("county_idx")
        .reindex(range(len(county_order)))
        .reset_index()
    )

    X = (
        week[neighbor_features]
        .fillna(0)
        .to_numpy(dtype=np.float32)
    )

    # Matrix multiplication
    neighbor_X = A @ X

    neighbor_df = pd.DataFrame(
        neighbor_X,
        columns=[
            f"neighbor_{c}"
            for c in neighbor_features
        ]
    )

    week = pd.concat(
        [
            week.reset_index(drop=True),
            neighbor_df
        ],
        axis=1
    )

    neighbor_tables.append(week)

weekly_df = pd.concat(
    neighbor_tables,
    ignore_index=True
)

# ------------------------------------------------------------
# Remove rows introduced by reindexing
# ------------------------------------------------------------

weekly_df = weekly_df[
    weekly_df["fips"].notna()
].copy()

print("Shape :", weekly_df.shape)

print()

print(
    weekly_df[
        [
            "score",
            "neighbor_score",
            "PRECTOT_mean_30d",
            "neighbor_PRECTOT_mean_30d"
        ]
    ].head()
)
```
